```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
df_all = pd.read_csv('screen time dataset countries/final_dataset_cleaned_data.csv')
df_all.head
```

```python
#creating a regression analysis to see if there is a correlation between screen time and country rank 
df_countryrank= pd.read_excel('WHR26_Data_Figure_2.1.xlsx')
countries_to_keep = [
    "United States of America",
    "Finland",
    "Iceland",
    "Denmark",
    "Germany",
    "France",
    "South Korea",
    "Japan",
    "India",
]

df_countryrank = df_countryrank[
    df_countryrank['Country name'].isin(countries_to_keep)
].copy()

df_countryrank = df_countryrank.rename(columns={'Country name': 'Country'})
df_countryrank = df_countryrank[['Country', 'Rank', 'Year']]
```

```python
df_countryrank = df_countryrank[
    df_countryrank['Year'].isin([2025, 2024, 2023, 2022, 2021, 2020, 2019, 2018, 2017, 2016, 2015])
]
```

```python
import statsmodels.formula.api as smf

df = df_all.merge(
    df_countryrank,
    left_on=["country_name", "year"],
    right_on=["Country", "Year"],
    how="inner"
)

print(df.shape, "rows after merge")

model = smf.ols(
    "Rank ~ screen_time_hours + addiction_score",
    data=df
)

result = model.fit(
    cov_type="cluster",
    cov_kwds={"groups": pd.factorize(df["country_name"])[0]}
)

print(result.summary())
```
